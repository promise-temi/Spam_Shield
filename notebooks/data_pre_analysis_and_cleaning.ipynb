```python
import os
import pandas as pd
```

```python
path = "../data/filtered_data/unprocessed/fr"

df_spam = pd.read_parquet(f'{path}/spam/df_filtered_spam_fr.parquet')
df_spam2 = pd.read_parquet(f'{path}/spam/sms_spam_detection_multilingual_spam_fr.parquet')
df_ham = pd.read_parquet(f'{path}/ham/df_filtered_ham_fr.parquet')
df_ham2 = pd.read_parquet(f'{path}/ham/sms_spam_detection_multilingual_ham_fr.parquet')
df_ham_pro = pd.read_parquet(f'{path}/ham/french_ham_contact_dataset_1000.parquet')
```

```python
df_ham = df_ham[df_ham.notna()]
df_ham['label'] = 0
df_spam = df_spam[df_spam.notna()]
df_ham_pro = df_ham_pro[df_ham_pro.notna()]
```

```python
df_all = pd.concat([df_ham_pro, df_ham,df_ham2, df_spam, df_spam2])

df_all = df_all.drop_duplicates(subset=['text'])
df_all[df_all['text'].duplicated()]
print(df_all.shape)
df_all.to_parquet(f'{path}/all/spam_ham_dataset.parquet')
```
