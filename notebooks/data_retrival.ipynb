Multiligual Spam Dataset

```python
# Download 
# Telechargement
import kagglehub
path = kagglehub.dataset_download("rajnathpatel/multilingual-spam-data", output_dir="../data/raw_data/multiligual_spam_dataset")
print("Path to dataset files:", path)
```

```python
# prudence pour eviter de saturer la mémoire mais il y a très peu de lignes juste 5572 (l'analyse statistitique s'arrete ici)
df = pd.read_csv("../data/raw_data/multiligual_spam_dataset/data-en-hi-de-fr.csv")
print(df.shape)
print(df.columns)
df.head(5)

 

```

```python


interesting_cols = ['labels','text', 'text_fr']
df_filtered = df[interesting_cols]

df_filtered['labels'] = df_filtered['labels'].replace('spam', 1)
df_filtered['labels'] = df_filtered['labels'].replace('ham', 0)


path = "../data/filtered_data/unprocessed"

# only the english ham messages
df_filtered_ham_eng = df_filtered[df_filtered['labels']==0][['labels', 'text']]
df_filtered_ham_eng = df_filtered_ham_eng.rename(columns={'labels':'label'})
df_filtered_ham_eng.to_parquet(f'{path}/eng/ham/df_filtered_ham_eng.parquet', engine="fastparquet")

# only the english spam messages
df_filtered_spam_eng = df_filtered[df_filtered['labels'] == 1][['labels', 'text']]
df_filtered_spam_eng = df_filtered_spam_eng.rename(columns={'labels':'label'})
df_filtered_spam_eng.to_parquet(f'{path}/eng/spam/df_filtered_spam_eng.parquet', engine="fastparquet")

# only the french ham messages
df_filtered_ham_fr = df_filtered[df_filtered['labels']==0][['labels', 'text_fr']]
df_filtered_ham_fr = df_filtered_ham_fr.rename(columns={'labels':'label', 'text_fr':'text'})
df_filtered_ham_fr.to_parquet(f'{path}/fr/ham/df_filtered_ham_fr.parquet', engine="fastparquet")

# only the french spam messages
df_filtered_spam_fr = df_filtered[df_filtered['labels'] == 1][['labels', 'text_fr']]
df_filtered_spam_fr = df_filtered_spam_fr.rename(columns={'labels':'label', 'text_fr':'text'})
df_filtered_spam_fr.to_parquet(f'{path}/fr/spam/df_filtered_spam_fr.parquet', engine="fastparquet")

```

Phishing Email Dataset

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("naserabdullahalam/phishing-email-dataset", output_dir="../data/raw_data/phishing_email_dataset")

print("Path to dataset files:", path)
```

```python
import os
# l'avantage ici est que les données sont triée par type de spam donc en quelque sorte prélabelisées
# df_csea = pd.read_csv('../data/raw_data/phishing_email_dataset/CEAS_08.csv')
path = "../data/filtered_data/unprocessed"
for file_name in os.listdir("../data/raw_data/phishing_email_dataset"):
    # print('.complete')
    if file_name == '.complete' :
        continue

    df = pd.read_csv(f'../data/raw_data/phishing_email_dataset/{file_name}')
    print(df.shape)
    print(df.columns)
    print(file_name)
    interesting_cols = ['label','body']
    df_filtered = df[interesting_cols]


    df_filtered_ham = df_filtered[df_filtered['label'] == 0]
    df_filtered_ham.rename(columns={'body':'text'})
    df_filtered_ham.to_parquet(f'{path}/eng/ham/{file_name.replace('.csv',"")}_filtered_ham.parquet', engine='fastparquet')


    df_filtered_spam = df_filtered[df_filtered['label'] == 1]
    df_filtered_spam.rename(columns={'body':'text'})
    df_filtered_spam.to_parquet(f'{path}/eng/spam/{file_name.replace('.csv',"")}_filtered_spam.parquet', engine='fastparquet')
```

# sms_spam_detection_multilingual

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("debapampal2002/sms-dataset1", output_dir="../data/raw_data/sms_spam_detection_multilingual")

print("Path to dataset files:", path)
```

```python
path = "../data/filtered_data/unprocessed"
df = pd.read_csv(f'../data/raw_data/sms_spam_detection_multilingual/dataset.csv')

df_fr_ham = df[(df['lang'] == 'french') & (df['labels'] == 'ham')][['labels', 'text']].replace({'ham':0}).rename(columns={'labels':'label'})
df_fr_ham.head()
df_fr_ham.to_parquet(f'{path}/fr/ham/sms_spam_detection_multilingual_ham_fr.parquet', engine="fastparquet")

df_fr_spam = df[(df['lang'] == 'french') & (df['labels'] == 'spam')][['labels', 'text']].replace({'spam':1}).rename(columns={'labels':'label'})
df_fr_spam.head()
df_fr_spam.to_parquet(f'{path}/fr/spam/sms_spam_detection_multilingual_spam_fr.parquet', engine="fastparquet")



df_eng_ham = df[(df['lang'] == 'english') & (df['labels'] == 'ham')][['labels', 'text']].replace({'ham':0}).rename(columns={'labels':'label'})
df_eng_ham.head()
df_eng_ham.to_parquet(f'{path}/fr/ham/sms_spam_detection_multilingual_ham_eng.parquet', engine="fastparquet")

df_eng_spam = df[(df['lang'] == 'english') & (df['labels'] == 'spam')][['labels', 'text']].replace({'spam':1}).rename(columns={'labels':'label'})
df_eng_spam.head()
df_eng_spam.to_parquet(f'{path}/fr/spam/sms_spam_detection_multilingual_spam_eng.parquet', engine="fastparquet")
```
