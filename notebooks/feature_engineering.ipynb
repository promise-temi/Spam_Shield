```python
import re

def remove_special_chars(text):

    # remplacer apostrophes et tirets par espace
    text = re.sub(r"[-']", " ", text)

    # supprimer autres caractères spéciaux
    text = re.sub(r"[^a-zA-ZÀ-ÿ0-9\s\[\]]", " ", text)

    # retirer espaces multiples
    text = re.sub(r"\s+", " ", text).strip()

    return text

df["text_final"] = df["text_transformed"].apply(remove_special_chars)
```

# NLTK REMOVE STOP WORDS

```python
def remove_stopwords(text):
    words = text.split()

    filtered_words = []

    for word in words:
        
        # garder les tokens
        if word.startswith("[") and word.endswith("]"):
            filtered_words.append(word)

        # retirer stopwords normaux
        elif word.lower() not in stop_words:
            filtered_words.append(word)

    return " ".join(filtered_words)
```

```python
df["text_final"] = df['text_final'].apply(remove_stopwords)
df["text_final"]
```

# NLTK STEMMING

```python
def stem_text(text):
    words = text.split()

    stemmed_words = []

    for word in words:

        # protéger tokens
        if word.startswith("[") and word.endswith("]"):
            stemmed_words.append(word)

        else:
            stemmed_words.append(stemmer.stem(word))

    return " ".join(stemmed_words)
```

```python
df["text_final"] = df['text_final'].apply(stem_text)
df["text_final"]
```

```python
def normalize_tokens(text):
    return re.sub(r"\[([a-zA-Z]+)\]", lambda m: "[" + m.group(1).upper() + "]", text)

def space_tokens(text):
    text = re.sub(r"(\[[A-Z]+\])", r" \1 ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text



df['text_final'] = df['text_final'].apply(normalize_tokens)
df['text_final'] = df['text_final'].apply(space_tokens)

df = df.replace([np.inf, -np.inf], np.nan).fillna(0)
df.columns
```
